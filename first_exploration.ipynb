Loaded a dataset and explored its shape and structure also found missing values. Analyzed survival rates by gender and class. Found statiscal measures like mean, min. max 

```python
import pandas as pd

df = pd.read_csv ('train.csv')

print(df.head(5))
print(df.tail(5))
print(df.info())
print(df.shape)
```

```python
print(df['Survived'].value_counts())
print(f"Average Age:{df['Age'].mean()}")
print(f"Average fare:{df['Fare'].mean()}")
```

```python
print(df['Sex'].value_counts())
print(df.groupby('Sex')['Survived'].mean())
print(df.groupby('Pclass')['Survived'].mean())
```

```python
#Prints the oldest passengers Age
print(f"Oldest Passenger: {df['Age'].max()}")

# Prints the youngest passengers Age
print(f"Youngest Passenger: {df['Age'].min()}")

# Prints how many vales are missing in the age column
print(f"Missing Age:{df['Age'].isnull().sum()}")

# Prints the most expensive fare
print(f"Fare Price: ${df['Fare'].max()}")
```
